### 1. Considerando el experimento de seleccionar un país participante al azar, se pide determinar las probabilidades de los siguientes sucesos:

a. pertenecer a África
b. pertenecer a Asia
c. pertenecer a Europa
d. pertenecer a Norteamérica
e. pertenecer a Sudamérica
f. clasificar a segunda ronda
g. ganar al menos un partido
h. pertenecer a África y clasificar a segunda ronda
i. pertenecer a Europa y clasificar a segunda ronda
j. pertenecer a Asia y clasificar a segunda ronda
k. pertenecer a Sudamérica y clasificar a segunda ronda
l. pertenecer a Norteamérica y clasificar a segunda ronda
m. pertenecer a África o clasificar a segunda ronda
n. pertenecer a Europa o clasificar a segunda ronda
o. pertenecer a Asia o clasificar a segunda ronda
p. pertenecer a Sudamérica o clasificar a segunda ronda
q. pertenecer a Norteamérica o clasificar a segunda ronda
r. pertenecer a África y haber ganado al menos un partido
s. pertenecer a Europa y haber ganado al menos un partido
t. pertenecer a Asia y haber ganado al menos un partido
u. pertenecer a Sudamérica y haber ganado al menos un partido
v. pertenecer a Norteamérica y haber ganado al menos un partido
w. haber ganado al menos un partido y clasificar a segunda ronda

```python
# Primero importamos la librería pandas y creamos el dataframe con el archivo csv.
import pandas as pd
cup2014= pd.read_csv("worldcup2014.csv",delimiter=',')

# Mostramos las primeras filas del dataframe creado.
cup2014.head()
```

```python
# Ahora, vamos a establecer algunas condiciones para establecer los conteos.

# Condiciones de continente.
africa = cup2014["continent"] == "africa"
asia = cup2014["continent"] == "asia"
europa = cup2014["continent"] == "europe"
norteamerica = cup2014["continent"] == "northamerica"
sudamerica = cup2014["continent"] == "southamerica"

# Condiciones de clasificación y juegos ganados.
segunda_ronda = cup2014["clasificado"] == 1
gano_partido = cup2014["juegos_ganados"] > 0

# Contamos la cantidad de equipos.
total = len(cup2014)
```

```python
# Vamos a crear un diccionario vacio para guardar los resultados y luego mostrarlos.
P = {}

# Realizamos los calculos de las probabilidades simples.
P["a_africa"] = africa.sum() / total
P["b_asia"] = asia.sum() / total
P["c_europa"] = europa.sum() / total
P["d_norteamerica"] = norteamerica.sum() / total
P["e_sudamerica"] = sudamerica.sum() / total
P["f_clasifica"] = segunda_ronda.sum() / total
P["g_gana_partido"] = gano_partido.sum() / total

# Probabilidades Conjuntas (Y).
P["h_africa_y_clasifica"] = (africa & segunda_ronda).sum() / total
P["i_europa_y_clasifica"] = (europa & segunda_ronda).sum() / total
P["j_asia_y_clasifica"] = (asia & segunda_ronda).sum() / total
P["k_sudamerica_y_clasifica"] = (sudamerica & segunda_ronda).sum() / total
P["l_norteamerica_y_clasifica"] = (norteamerica & segunda_ronda).sum() / total

# Probabilidades con Uniones (O).
P["m_africa_o_clasifica"] = (africa | segunda_ronda).sum() / total
P["n_europa_o_clasifica"] = (europa | segunda_ronda).sum() / total
P["o_asia_o_clasifica"] = (asia | segunda_ronda).sum() / total
P["p_sudamerica_o_clasifica"] = (sudamerica | segunda_ronda).sum() / total
P["q_norteamerica_o_clasifica"] = (norteamerica | segunda_ronda).sum() / total

# Probabilidades conjuntas con ganar partido.
P["r_africa_y_gana"] = (africa & gano_partido).sum() / total
P["s_europa_y_gana"] = (europa & gano_partido).sum() / total
P["t_asia_y_gana"] = (asia & gano_partido).sum() / total
P["u_sudamerica_y_gana"] = (sudamerica & gano_partido).sum() / total
P["v_norteamerica_y_gana"] = (norteamerica & gano_partido).sum() / total

# Probabilidades ganar partido y clasificar.
P["w_gana_y_clasifica"] = (gano_partido & segunda_ronda).sum() / total


# Imprimimos un título.
print("\n--- Resultados del Punto 1: Cálculo de Probabilidades ---\n")

# Mostramos los resultados.
for clave, valor in P.items():
    print(f"{clave}: {valor:.2%}")
```

### 2. A partir de las probabilidades anteriores, y sin realizar conteos, calcular las siguientes probabilidades:

```python
a. pertenecer a Europa, si se sabe que clasificó
b. pertenecer a Sudamérica, si se sabe que clasificó
c. pertenecer a África, si se sabe que ganó al menos un partido
d. haber ganado al menos un partido, si se sabe que clasificó
e. haber ganado al menos un partido, si se sabe que pertenece a Norteamérica
f. NO haber clasificado, si se sabe que pertenece a África
```

```python
# Probabilidades condicionales usando la fórmula:
# P(A | B) = P(A ∩ B) / P(B)

# a. P(Europa | Clasifica).
P_europa_dado_clasifica = P["i_europa_y_clasifica"] / P["f_clasifica"]

# b. P(Sudamérica | Clasifica).
P_sudamerica_dado_clasifica = P["k_sudamerica_y_clasifica"] / P["f_clasifica"]

# c. P(África | Gana al menos un partido).
P_africa_dado_gana = P["r_africa_y_gana"] / P["g_gana_partido"]

# d. P(Gana al menos un partido | Clasifica).
P_gana_dado_clasifica = P["w_gana_y_clasifica"] / P["f_clasifica"]

# e. P(Gana al menos un partido | Norteamérica).
P_gana_dado_norteamerica = P["v_norteamerica_y_gana"] / P["d_norteamerica"]

# f. P(NO clasifica | África).
# Como en el punto anterior no calculamos directamente “África y no clasifica".
# Tenemos que todos los equipos africanos se dividen en dos grupos:los que clasifican y los que NO clasifican.
# Entonces: 
# P(Africa)=P(Africa ∩ Clasifica)+P(Africa ∩ No clasifica)
# Despejamos lo que nos interesa (P(Africa ∩ No clasifica)): 
# P(Africa ∩ No clasifica)=P(Africa)−P(Africa ∩ Clasifica)
# Reemplazamos con los valores y luego calculamos la probabilidad condicional.
P_no_clasifica_y_africa = P["a_africa"] - P["h_africa_y_clasifica"]
P_no_clasifica_dado_africa = P_no_clasifica_y_africa / P["a_africa"]

# Creamos un diccionario con las probabilidades del punto 2, para poder presentar mejor los resultados.
P2 = {
    "a) P(Europa | Clasifica)": P_europa_dado_clasifica,
    "b) P(Sudamérica | Clasifica)": P_sudamerica_dado_clasifica,
    "c) P(África | Gana un partido)": P_africa_dado_gana,
    "d) P(Gana un partido | Clasifica)": P_gana_dado_clasifica,
    "e) P(Gana un partido | Norteamérica)": P_gana_dado_norteamerica,
    "f) P(No clasifica | África)": P_no_clasifica_dado_africa
}

# Imprimimos un título.
print("\n--- Resultados del Punto 2: Cálculo de Probabilidades Condicionales ---\n")

# Mostramos los resultados.
for clave, valor in P2.items():
    print(f"{clave}: {valor:.2%}")
```
